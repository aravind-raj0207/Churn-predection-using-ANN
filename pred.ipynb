```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd
import pickle
import numpy as np
```

```python
model=load_model("model.h5")

with open("lb_gender.pk1",'rb') as file:
    lb_gender=pickle.load(file)

with open("oh_geo.pk2",'rb') as file:
    oh_geo=pickle.load(file)
    
with open("scalar",'rb') as file:
    scalar=pickle.load(file)
```

```python
input={'CreditScore':600,'Geography':'France' ,'Gender':'Male', 'Age':40, 'Tenure':3, 'Balance':6000, 'NumOfProducts':2,
       'HasCrCard':1, 'IsActiveMember':1, 'EstimatedSalary':50000}
```

```python
geo_encoder=oh_geo.transform([[input["Geography"]]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoder,columns=oh_geo.get_feature_names_out(["Geography"]))
```

```python
geo_encoded_df
```

```python
input_df=pd.DataFrame([input])
input_df
```

```python
input_df["Gender"]=lb_gender.transform(input_df["Gender"])
```

```python
input_df=pd.concat([input_df.drop("Geography",axis=1),geo_encoded_df],axis=1)
```

```python
input_df
```

```python
input_scaled=scalar.transform(input_df)
input_scaled
```

```python
predection=model.predict(input_scaled)
predection
```

```python
predection_prob=predection[0][0]
```

```python
predection_prob
```

```python
if predection_prob > 0.5:
    print("the customer is likely to churn")
else:
    print("the customer is not likely to churn")
```
